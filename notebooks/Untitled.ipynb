```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
```

```python
with rasterio.open('../scratch/glofas_brussels_forecast.tif') as f:
    x = f.read()
    
```

```python
x.shape
```

```python
a,b = np.percentile(x[~np.isnan(x)],(1,99))
fig, axs = plt.subplots(1,len(x), figsize=(len(x)*4, 4))
axs = axs.flatten()
for i in range(len(x)):
    ax = axs[i]
    ax.imshow(x[i], vmin=a, vmax=b)
```
